# MLP: capacity vs dropout / L2

Same noisy 1D function, three nets. Overfit, then regularize.


```python
import numpy as np
import matplotlib.pyplot as plt

from interview_prep.metrics import mse
from interview_prep.models import MLP

rng = np.random.default_rng(0)
X = rng.uniform(-3, 3, size=(80, 1))
y = np.sin(1.7 * X.ravel()) + rng.normal(scale=0.25, size=80)
grid = np.linspace(-3, 3, 200).reshape(-1, 1)
truth = np.sin(1.7 * grid.ravel())

def fit(name, **kw):
    net = MLP(hidden=64, lr=0.08, max_iter=2500, random_state=0, **kw).fit(X, y)
    print(f"{name:16} train MSE={mse(y, net.predict(X)):.3f}")
    return net

plain = fit("no reg")
l2 = fit("l2=0.02", l2=0.02)
drop = fit("dropout=0.3", dropout=0.3)

plt.figure(figsize=(8, 4.5))
plt.scatter(X.ravel(), y, s=18, alpha=0.7, label="train")
plt.plot(grid, truth, color="black", lw=2, label="sin")
plt.plot(grid, plain.predict(grid), label="MLP")
plt.plot(grid, l2.predict(grid), label="MLP + L2")
plt.plot(grid, drop.predict(grid), label="MLP + dropout")
plt.legend()
plt.title("Regularization vs a small noisy dataset")
plt.show()
```
